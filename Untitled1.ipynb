```python

import os
import gc
import torch
import fitz
import faiss
import numpy as np

from google.colab import files
from sentence_transformers import SentenceTransformer

from unsloth import FastLanguageModel

print("All libraries imported successfully!")
```

```python

if torch.cuda.is_available():

    print("GPU Available:", torch.cuda.get_device_name(0))

    total_memory = torch.cuda.get_device_properties(0).total_memory

    print("Total GPU Memory:",
          round(total_memory / (1024**3), 2), "GB")

else:
    raise RuntimeError(
        "GPU not available. Go to Runtime > Change runtime type > T4 GPU."
    )
```

```python

MODEL_NAME = "unsloth/Qwen2.5-1.5B-Instruct-unsloth-bnb-4bit"

MAX_SEQ_LENGTH = 2048
LOAD_IN_4BIT = True

print("Model:", MODEL_NAME)
print("4-bit Quantization:", LOAD_IN_4BIT)
```

```python

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name=MODEL_NAME,
    max_seq_length=MAX_SEQ_LENGTH,
    dtype=None,
    load_in_4bit=True
)

FastLanguageModel.for_inference(model)

print("Unsloth model loaded successfully!")
```

```python

torch.cuda.synchronize()

allocated = torch.cuda.memory_allocated() / (1024**3)
reserved = torch.cuda.memory_reserved() / (1024**3)

print(f"Allocated GPU Memory: {allocated:.2f} GB")
print(f"Reserved GPU Memory: {reserved:.2f} GB")
```

```python

uploaded_files = files.upload()

print("\nUploaded Files:")

for filename in uploaded_files.keys():
    print(filename)
```

```python

documents = []

for filename in uploaded_files.keys():

    if filename.lower().endswith(".pdf"):

        pdf = fitz.open(filename)

        for page_number, page in enumerate(pdf):

            text = page.get_text()

            if text.strip():

                documents.append({
                    "text": text,
                    "source": filename,
                    "page": page_number + 1
                })

        pdf.close()

    elif filename.lower().endswith(".txt"):

        with open(filename, "r", encoding="utf-8") as file:

            text = file.read()

        if text.strip():

            documents.append({
                "text": text,
                "source": filename,
                "page": None
            })

print("Documents extracted:", len(documents))
```

```python

for doc in documents[:3]:

    print("Source:", doc["source"])
    print("Page:", doc["page"])
    print("Content:")
    print(doc["text"][:500])
    print("-" * 60)
```

```python

def split_text(text, chunk_size=400, overlap=80):

    words = text.split()
    chunks = []

    start = 0

    while start < len(words):

        end = start + chunk_size

        chunk = " ".join(words[start:end])

        if chunk.strip():
            chunks.append(chunk)

        start += chunk_size - overlap

    return chunks


print("Text chunking function created!")
```

```python

chunks = []
chunk_metadata = []

for doc in documents:

    text_chunks = split_text(doc["text"])

    for chunk in text_chunks:

        chunks.append(chunk)

        chunk_metadata.append({
            "source": doc["source"],
            "page": doc["page"]
        })

print("Total Chunks Created:", len(chunks))
```

```python

for i, chunk in enumerate(chunks[:3]):

    print(f"Chunk {i+1}")
    print(chunk[:400])
    print("-" * 60)
```

```python

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2",
    device="cpu"
)

print("Embedding model loaded successfully!")
```

```python

embeddings = embedding_model.encode(
    chunks,
    batch_size=32,
    show_progress_bar=True,
    convert_to_numpy=True,
    normalize_embeddings=True
)

embeddings = embeddings.astype("float32")

print("Embeddings Generated Successfully!")
print("Embedding Shape:", embeddings.shape)
```

```python

dimension = embeddings.shape[1]

index = faiss.IndexFlatIP(dimension)

index.add(embeddings)

print("FAISS Index Created Successfully!")
print("Total Indexed Vectors:", index.ntotal)
```

```python

print("Vector Database Status")

print("Total Vectors:", index.ntotal)
print("Embedding Dimensions:", dimension)
print("Database Ready:", index.ntotal > 0)
```

```python

def retrieve_documents(query, top_k=3):

    if index.ntotal == 0:
        return []

    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype("float32")

    k = min(top_k, index.ntotal)

    scores, indices = index.search(
        query_embedding, k
    )

    results = []

    for score, idx in zip(scores[0], indices[0]):

        if idx == -1:
            continue

        results.append({
            "text": chunks[idx],
            "source": chunk_metadata[idx]["source"],
            "page": chunk_metadata[idx]["page"],
            "score": float(score)
        })

    return results


print("Retrieval Function Created Successfully!")
```

```python

query = "What is the main topic of this document?"

results = retrieve_documents(query)

for i, result in enumerate(results):

    print(f"Result {i+1}")
    print("Source:", result["source"])
    print("Page:", result["page"])
    print("Similarity Score:", round(result["score"], 3))
    print("Text:", result["text"][:400])
    print("-" * 60)
```

```python

def generate_rag_answer(query):

    retrieved_docs = retrieve_documents(query, top_k=3)

    if not retrieved_docs:

        return "No relevant documents found.", []

    context = "\n\n".join([
        f"Source: {doc['source']}, Page: {doc['page']}\n"
        f"{doc['text']}"
        for doc in retrieved_docs
    ])

    messages = [
        {
            "role": "system",
            "content": (
                "You are a document-based question answering assistant. "
                "Answer using only the provided context. "
                "If the answer is not present in the context, "
                "say that the documents do not contain enough information. "
                "Do not invent facts."
            )
        },
        {
            "role": "user",
            "content": (
                f"Context:\n{context}\n\n"
                f"Question: {query}\n\n"
                "Provide a clear and accurate answer."
            )
        }
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt"
    ).to("cuda")

    with torch.inference_mode():

        outputs = model.generate(
            input_ids=inputs,
            max_new_tokens=200,
            do_sample=False,
            use_cache=True,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_tokens = outputs[0][inputs.shape[-1]:]

    answer = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return answer.strip(), retrieved_docs


print("RAG Generation Function Created!")
```

```python

question = "What is the main topic discussed in the uploaded documents?"

answer, sources = generate_rag_answer(question)

print("USER QUESTION:")
print(question)

print("\nGENERATED ANSWER:")
print(answer)

print("\nRETRIEVED SOURCES:")

for source in sources:

    print("File:", source["source"])
    print("Page:", source["page"])
    print("Similarity:", round(source["score"], 3))
    print("-" * 50)
```

```python

while True:

    question = input("\nEnter your question (type exit to stop): ")

    if question.lower() == "exit":
        print("RAG system stopped!")
        break

    answer, sources = generate_rag_answer(question)

    print("\nANSWER:")
    print(answer)

    print("\nSOURCES:")

    for source in sources:

        print(
            f"File: {source['source']} | "
            f"Page: {source['page']}"
        )

    print("-" * 60)
```
